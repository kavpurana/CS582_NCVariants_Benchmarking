Prepare 1kb windows for DeepSEA since it only takes a 1000 bp sequence that has the variant centered in the middle 500 position. There are two sequences per variant: the reference and the alternate

```python
#Helper functions
def extract_centered_sequence(fasta, chrom, pos, flank=500):
    """
    Extract 1001 bp window centered on 1-based genomic position.
    Returns sequence as uppercase string.
    """
    start = pos - flank - 1        # 1-based → 0-based
    end   = pos + flank            # inclusive of pos+500

    seq = fasta[chrom][start:end].seq.upper()

    if len(seq) != flank*2 + 1:
        return None  # skip variants near chromosome edges

    return seq

def normalize_ref_alt(seq, ref, alt):
    """
    Ensures ref matches the center base of seq.
    Performs strand flip if needed.
    seq length must be odd → SNP index = len(seq)//2
    """
    ref = ref.upper()
    alt = alt.upper()

    snp_index = len(seq) // 2
    genome_base = seq[snp_index]

    # direct match
    if genome_base == ref:
        return ref, alt

    # reverse complement match
    if genome_base == str(Seq(ref).complement()):
        print("Strand mismatch detected → flipping REF/ALT")
        return (
            str(Seq(ref).complement()).upper(),
            str(Seq(alt).complement()).upper()
        )

    # neither matches → invalid variant
    return None, None


def apply_mutation(seq, ref, alt):
    """
    Returns the sequence with the SNP substituted at the center.
    """
    snp_index = len(seq) // 2
    assert seq[snp_index] == ref, f"REF mismatch: expected {ref}, got {seq[snp_index]}"

    return seq[:snp_index] + alt + seq[snp_index+1:]


def preprocess_variants(df):
    records = []

    for i, row in df.iterrows():
        chrom = row["chr_hg19"].replace("chr", "")
        pos   = int(row["pos_hg19"])
        ref   = row["ref"]
        alt   = row["alt"]
        label = row["ID"]

        # 1. Extract sequence
        seq_ref = extract_centered_sequence(hg19, chrom, pos)
        if seq_ref is None:
            print(f"Skipping {label}: near chromosome boundary")
            continue

        # 2. Normalize alleles
        ref_norm, alt_norm = normalize_ref_alt(seq_ref, ref, alt)
        if ref_norm is None:
            print(f"Skipping {label}: REF does not match hg19 ({ref} vs {seq_ref[len(seq_ref)//2]})")
            continue

        # 3. Create ALT-mutated sequence
        try:
            seq_alt = apply_mutation(seq_ref, ref_norm, alt_norm)
        except AssertionError as e:
            print(f"Skipping {label}: {e}")
            continue

        # 4. Store results
        records.append({
            "id": label,
            "chrom": chrom,
            "pos": pos,
            "ref": ref_norm,
            "alt": alt_norm,
            "seq_ref": seq_ref,
            "seq_alt": seq_alt
        })

    return pd.DataFrame(records)
```

# Enformer Pathogenic Data Preparation

I manually changed the VCF file from CHR to CHROM

```python
from cyvcf2 import VCF
from pyfaidx import Fasta
import numpy as np

vcf = VCF("/content/ncVarDB/data/ncVar_pathogenic.vcf")
fasta = Fasta("hg38.fa")
```

```python
SEQUENCE_LENGTH = 393_216  # Input sequence length in base pairs
TARGET_LENGTH = 896        # Number of 128bp bins predicted
NUM_TRACKS_HUMAN = 5313    # Number of human tracks
HALF_WINDOW = SEQUENCE_LENGTH // 2 #196,608
```

```python
variants = []

for var in vcf:
    chrom = var.CHROM
    pos = var.POS
    ref = var.REF
    alt = var.ALT[0]

    if not chrom.startswith("chr"):
        chrom = "chr" + chrom

    # Extract full Enformer window
    start = pos - HALF_WINDOW
    end = pos + HALF_WINDOW
    # Skip variants near chromosome boundaries
    if start < 1 or end > len(fasta[chrom]):
      print(f"Skipping {chrom}:{pos} — near chromosome edge")
      continue

    seq = fasta[chrom][start:end].seq.upper()

    # Skip if near chromosome edges
    if len(seq) != SEQUENCE_LENGTH:
        print(f"Skipping {chrom}:{pos} — near chromosome edge ({len(seq)} bp)")
        continue

    # Index of variant within the 393k sequence
    center_index = HALF_WINDOW

    # REF allele check
    if seq[center_index] != ref:
        print(f"Skipping {chrom}:{pos} — REF mismatch ({seq[center_index]} vs {ref})")
        continue

    # Build REF/ALT sequences
    ref_seq = seq
    alt_seq = seq[:center_index] + alt + seq[center_index+1:]

    variants.append({
        "id": var.ID,
        "chrom": chrom,
        "pos": pos,
        "ref": ref,
        "alt": alt,
        "ref_seq": ref_seq,   # <-- REQUIRED
        "alt_seq": alt_seq   # <-- REQUIRED
    })
```

# Testing Enformer Predictions on Pathogenic

```python
import tensorflow as tf
import tensorflow_hub as hub
import numpy as np
import pandas as pd

# load DeepMind pretrained Enformer
enformer_model = hub.load('https://tfhub.dev/deepmind/enformer/1').model

print("Loaded Enformer successfully!")
```

```python
def one_hot_encode(sequence):
    """
    One-hot encode DNA sequence.

    Args:
        sequence: String of DNA bases (A, C, G, T, N)

    Returns:
        One-hot encoded array of shape (len(sequence), 4)
        Order: A, C, G, T
    """
    sequence = sequence.upper()
    vocab = {'A': 0, 'C': 1, 'G': 2, 'T': 3}

    # Initialize with zeros (N bases will be all zeros)
    one_hot = np.zeros((len(sequence), 4), dtype=np.float32)

    for i, base in enumerate(sequence):
        if base in vocab:
            one_hot[i, vocab[base]] = 1

    return one_hot


def predict_on_sequence(sequence):
    """
    Make prediction on a DNA sequence.

    Args:
        sequence: DNA sequence string (must be 393,216 bp)

    Returns:
        Dictionary with 'human' predictions of shape (896, 5313)
    """
    if len(sequence) != SEQUENCE_LENGTH:
        raise ValueError(f"Sequence must be {SEQUENCE_LENGTH} bp, got {len(sequence)}")

    # One-hot encode
    seq_one_hot = one_hot_encode(sequence)

    # Add batch dimension
    seq_batch = seq_one_hot[np.newaxis, ...]

    # Convert to TensorFlow tensor with correct dtype
    seq_tensor = tf.constant(seq_batch, dtype=tf.float32)

    # Make prediction
    predictions = enformer_model.predict_on_batch(seq_tensor)

    return predictions


def compute_variant_effect(ref_sequence, alt_sequence):
    """
    Compute effect of a variant by comparing predictions.

    Args:
        ref_sequence: Reference sequence (393,216 bp)
        alt_sequence: Alternate sequence with variant (393,216 bp)

    Returns:
        Dictionary with SAD scores and predictions
    """

    for v in variants:
      # Get predictions
      ref_pred = predict_on_sequence(ref_sequence)
      alt_pred = predict_on_sequence(alt_sequence)

      # Extract human tracks
      ref_human = ref_pred['human'][0]  # Remove batch dimension
      alt_human = alt_pred['human'][0]

      # Compute SNP Activity Difference (SAD)
      sad = alt_human - ref_human

      # Compute SNP Activity Ratio (SAR) - log2 fold change
      sar = np.log2((1 + alt_human) / (1 + ref_human))

      return {
          'sad': sad,  # Shape: (896, 5313)
          'sar': sar,  # Shape: (896, 5313)
          'ref_prediction': ref_human,
          'alt_prediction': alt_human
      }
```

```python
variant_results = []
save_sad = False   # set True if you want full SAD matrices (VERY large)

print("\nRunning Enformer on all variants...\n")

for i, v in enumerate(variants):
    print(f"Processing {i+1}/{len(variants)} → {v['chrom']}:{v['pos']} {v['ref']}>{v['alt']}")

    ref_seq = v["ref_seq"]
    alt_seq = v["alt_seq"]

    # Compute variant effect
    ve = compute_variant_effect(ref_seq, alt_seq)

    # Reduce SAD to a single numeric score
    sad = ve["sad"]
    sar = ve["sar"]

    # Reduce SAD to single pathogenicity score
    score = np.max(np.abs(sad))

    # Store result
    variant_results.append({
        "id": v["id"],
        "chrom": v["chrom"],
        "pos": v["pos"],
        "ref": v["ref"],
        "alt": v["alt"],
        "score": score
    })


print("\n✓ Completed Enformer inference for all variants!")
```

```python
import pandas as pd

df = pd.DataFrame(variant_results)
df.to_csv("enformer_variant_scores.csv", index=False)

print("Saved to enformer_variant_scores.csv!")
```
